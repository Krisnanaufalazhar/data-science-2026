## Deskripsi Praktikum

Pada praktikum ini dilakukan pembersihan data (data cleaning) sederhana dan pembuatan dashboard visualisasi menggunakan dataset Titanic. Dashboard dibuat menggunakan Matplotlib dan Seaborn untuk menampilkan berbagai visualisasi seperti bar chart, histogram, boxplot, dan scatter plot guna memahami karakteristik data serta hubungan antar variabel.


# IMPORT LIBRARY

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
```

# LOAD DATASET

```python
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

print(df.shape)
print(df.dtypes)
print(df.head())
```

# DATA CLEANING SEDERHANA

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

# SET STYLE

```python
sns.set_style("whitegrid")
```

# SET UP DASHBOARD

```python
fig = plt.figure(figsize=(16,12))
fig.suptitle(
    'Dashboard Analisis Dataset Titanic',
    fontsize=20,
    fontweight='bold',
    y=0.98
)

gs = gridspec.GridSpec(
    2, 2,
    figure=fig,
    hspace=0.35,
    wspace=0.25
)

ax1 = fig.add_subplot(gs[0,0])
ax2 = fig.add_subplot(gs[0,1])
ax3 = fig.add_subplot(gs[1,0])
ax4 = fig.add_subplot(gs[1,1])

# GRAFIK 1 : BAR CHART

survival_class = df.groupby('Pclass')['Survived'].mean() * 100

survival_class.plot(
    kind='bar',
    ax=ax1,
    color=['#1f77b4','#ff7f0e','#2ca02c'],
    edgecolor='black'
)

ax1.set_title('Persentase Survival Berdasarkan Kelas')
ax1.set_xlabel('Kelas Penumpang')
ax1.set_ylabel('Persentase Selamat (%)')
ax1.tick_params(axis='x', rotation=0)
ax1.spines[['top','right']].set_visible(False)

# GRAFIK 2 : HISTOGRAM

sns.histplot(
    data=df,
    x='Age',
    hue='Survived',
    kde=True,
    bins=20,
    palette='Set2',
    ax=ax2
)

ax2.set_title('Distribusi Umur Penumpang')
ax2.set_xlabel('Umur')
ax2.set_ylabel('Jumlah Penumpang')
ax2.spines[['top','right']].set_visible(False)

# GRAFIK 3 : BOXPLOT

sns.boxplot(
    data=df,
    x='Sex',
    y='Fare',
    palette='Set3',
    ax=ax3
)

ax3.set_title('Distribusi Tarif Berdasarkan Gender')
ax3.set_xlabel('Gender')
ax3.set_ylabel('Fare')
ax3.spines[['top','right']].set_visible(False)


# GRAFIK 4 : SCATTER PLOT

sns.scatterplot(
    data=df,
    x='Age',
    y='Fare',
    hue='Survived',
    size='Pclass',
    sizes=(40,200),
    alpha=0.7,
    palette='Set1',
    ax=ax4
)

ax4.set_title('Hubungan Umur dan Tarif Penumpang')
ax4.set_xlabel('Umur')
ax4.set_ylabel('Fare')
ax4.spines[['top','right']].set_visible(False)


# EXPORT DASHBOARD


plt.savefig(
    'dashboard_titanic.png',
    dpi=200,
    bbox_inches='tight'
)

plt.show()

print("Dashboard berhasil disimpan sebagai dashboard_titanic.png")
```
